# 02 — Lists, Dictionaries, and Loops Practice

This notebook practices three basic Python tools that appear constantly in Machine Learning: lists, dictionaries, and loops.

## 1. Store Experiment Results

A list of dictionaries is a simple way to store many experiment results. Each dictionary represents one experiment.

```python
experiments = [
    {"model": "KNN", "accuracy": 0.82},
    {"model": "Logistic Regression", "accuracy": 0.86},
    {"model": "Random Forest", "accuracy": 0.89},
    {"model": "SVM", "accuracy": 0.84},
]

experiments
```

## 2. Filter Good Experiments

Now I can loop through experiments and print only the models with accuracy greater than 0.85.

```python
for experiment in experiments:
    if experiment["accuracy"] > 0.85:
        print(experiment["model"], experiment["accuracy"])
```

## 3. Find the Best Experiment

The best model is the one with the highest accuracy in this small example.

```python
best_experiment = experiments[0]

for experiment in experiments:
    if experiment["accuracy"] > best_experiment["accuracy"]:
        best_experiment = experiment

best_experiment
```

## 4. Manual Accuracy Calculation

Accuracy is the number of correct predictions divided by the total number of predictions.

$$
\mathrm{Accuracy} = \frac{\text{number of correct predictions}}{\text{total number of predictions}}
$$

```python
true_labels = [1, 0, 1, 1, 0]
predicted_labels = [1, 0, 0, 1, 0]

correct = 0

for i in range(len(true_labels)):
    if true_labels[i] == predicted_labels[i]:
        correct = correct + 1

accuracy = correct / len(true_labels)

print("Correct predictions:", correct)
print("Accuracy:", accuracy)
```
