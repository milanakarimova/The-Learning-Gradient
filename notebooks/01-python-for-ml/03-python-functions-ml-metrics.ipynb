# 03 — Python Functions for ML Metrics

This notebook practices writing reusable Python functions for simple Machine Learning-style metrics.

## 1. Mean Function

The mean is the average of values.

$$
\bar{x} = \frac{1}{n}\sum_{i=1}^{n}x_i
$$

```python
def calculate_mean(values):
    return sum(values) / len(values)

scores = [0.72, 0.75, 0.80, 0.82]
calculate_mean(scores)
```

## 2. Accuracy Function

Accuracy is the number of correct predictions divided by the total number of predictions.

$$
\mathrm{Accuracy} = \frac{\text{number of correct predictions}}{\text{total number of predictions}}
$$

```python
def calculate_accuracy(true_labels, predicted_labels):
    correct = 0

    for i in range(len(true_labels)):
        if true_labels[i] == predicted_labels[i]:
            correct = correct + 1

    return correct / len(true_labels)

true_labels = [1, 0, 1, 1, 0]
predicted_labels = [1, 0, 0, 1, 0]

calculate_accuracy(true_labels, predicted_labels)
```

## 3. MSE Function

Mean Squared Error measures the average squared difference between true values and predicted values.

$$
\mathrm{MSE} = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2
$$

```python
def calculate_mse(y_true, y_pred):
    total_error = 0

    for i in range(len(y_true)):
        error = y_true[i] - y_pred[i]
        squared_error = error ** 2
        total_error = total_error + squared_error

    return total_error / len(y_true)

y_true = [3, 5, 2, 7]
y_pred = [2.5, 5.5, 2, 8]

calculate_mse(y_true, y_pred)
```
