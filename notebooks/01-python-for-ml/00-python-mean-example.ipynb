# 00 — Python Mean Example

This notebook is a small first experiment for the Python for Machine Learning section.

The goal is simple: take a list of numbers and calculate their mean. Even though this example is small, it already shows an important Machine Learning habit: we start from data, calculate something useful, and interpret the result.

## 1. The Idea

The mean is the average value of a group of numbers. Mathematically, it is written as:

$$
\bar{x} = \frac{1}{n}\sum_{i=1}^{n}x_i
$$

In simple words, we add all numbers together and divide by how many numbers we have.

````python
numbers = [10, 20, 30, 40, 50]

mean = sum(numbers) / len(numbers)

print("Numbers:", numbers)
print("Mean:", mean)
````

## 2. What Happened?

The list `numbers` stores the data. The function `sum(numbers)` adds all values. The function `len(numbers)` counts how many values exist in the list.

So this line:

```python
mean = sum(numbers) / len(numbers)
```

is the Python version of the mathematical formula above.

````python
total = sum(numbers)
count = len(numbers)

print("Total:", total)
print("Count:", count)
print("Mean:", total / count)
````

## 3. Small Reflection

This example is simple, but it teaches a useful pattern:

```text
data -> calculation -> result -> interpretation
```

This pattern will appear again and again in Machine Learning.
