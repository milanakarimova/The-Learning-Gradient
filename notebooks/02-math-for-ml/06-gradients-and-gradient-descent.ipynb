# 06 — Gradients and Gradient Descent Practice

This notebook practices gradient descent for linear regression.

```python
import numpy as np
```

## 1. Prediction and MSE

$$
\hat{y}=Xw+b
$$

$$
\mathrm{MSE}=\frac{1}{n}\sum_i(y_i-\hat{y}_i)^2
$$

```python
def predict(X, w, b):
    return X @ w + b


def mse(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)
```

## 2. Vectorized Gradients

$$
\nabla_w \mathcal{L}= -\frac{2}{n}X^T(y-\hat{y})
$$

$$
\frac{\partial \mathcal{L}}{\partial b}= -\frac{2}{n}\sum_i(y_i-\hat{y}_i)
$$

```python
def gradients(X, y, w, b):
    n = X.shape[0]
    y_pred = predict(X, w, b)
    errors = y - y_pred
    dL_dw = -(2 / n) * (X.T @ errors)
    dL_db = -(2 / n) * np.sum(errors)
    return dL_dw, dL_db
```

## 3. Gradient Descent Training Loop

$$
\theta_{new}=\theta_{old}-\alpha\nabla_\theta\mathcal{L}
$$

```python
def train_gradient_descent(X, y, learning_rate=0.01, steps=1000):
    n_features = X.shape[1]
    w = np.zeros(n_features)
    b = 0.0
    losses = []

    for step in range(steps + 1):
        y_pred = predict(X, w, b)
        loss = mse(y, y_pred)
        losses.append(loss)

        dL_dw, dL_db = gradients(X, y, w, b)
        w = w - learning_rate * dL_dw
        b = b - learning_rate * dL_db

    return w, b, losses
```

## 4. Train on a Simple Dataset

```python
x = np.array([1, 2, 3, 4, 5], dtype=float)
y = 2 * x + 1
X = x.reshape(-1, 1)

w, b, losses = train_gradient_descent(X, y, learning_rate=0.01, steps=1000)

w, b, losses[-1]
```
