# 11 — Eigenvectors, PCA, and Dimensionality Reduction Practice

This notebook practices PCA from scratch using NumPy: centering, covariance, eigen-decomposition, projection, reconstruction, explained variance, and scaling effects.

```python
import numpy as np
```

## 1. Create a Correlated Dataset

```python
rng = np.random.default_rng(42)

mean = np.array([0, 0])
covariance_true = np.array([
    [3.0, 1.8],
    [1.8, 1.4],
])

X = rng.multivariate_normal(mean, covariance_true, size=300)
X.shape
```

## 2. Center and Compute Covariance

$$
X_c = X - \mu
$$

$$
\Sigma=\frac{1}{n-1}X_c^TX_c
$$

```python
mu = X.mean(axis=0)
X_centered = X - mu
covariance = (X_centered.T @ X_centered) / (X.shape[0] - 1)

mu, covariance
```

## 3. Eigen-Decomposition

$$
\Sigma v = \lambda v
$$

```python
eigenvalues, eigenvectors = np.linalg.eigh(covariance)

order = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[order]
eigenvectors = eigenvectors[:, order]

eigenvalues, eigenvectors
```

## 4. Projection onto First Principal Component

$$
Z = X_c W_k
$$

```python
W1 = eigenvectors[:, :1]
Z = X_centered @ W1

X.shape, Z.shape
```

## 5. Reconstruction

$$
\hat{X}=ZW_k^T+\mu
$$

```python
X_reconstructed = Z @ W1.T + mu
reconstruction_error = np.mean(np.sum((X - X_reconstructed) ** 2, axis=1))

reconstruction_error
```

## 6. Explained Variance Ratio

$$
\frac{\lambda_j}{\sum_m \lambda_m}
$$

```python
explained_variance_ratio = eigenvalues / eigenvalues.sum()
explained_variance_ratio
```

## 7. Scaling Effect

```python
X_scaled_problem = X.copy()
X_scaled_problem[:, 1] *= 20

X_standardized = (X_scaled_problem - X_scaled_problem.mean(axis=0)) / X_scaled_problem.std(axis=0)

cov_unscaled = np.cov(X_scaled_problem.T)
cov_standardized = np.cov(X_standardized.T)

vals_unscaled, vecs_unscaled = np.linalg.eigh(cov_unscaled)
vals_standardized, vecs_standardized = np.linalg.eigh(cov_standardized)

np.sort(vals_unscaled)[::-1] / vals_unscaled.sum(), np.sort(vals_standardized)[::-1] / vals_standardized.sum()
```

## Reflection

PCA finds the directions of maximum variance and projects data into a lower-dimensional coordinate system. It is linear algebra, statistics, and geometry working together.
