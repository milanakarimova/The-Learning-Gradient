# 05 — Functions, Slopes, and Derivatives Practice

This notebook practices exact derivatives, numerical derivatives, MSE derivatives, and one manual gradient descent update.

```python
import numpy as np
```

## 1. Exact and Numerical Derivatives

For $f(x)=x^2$:

$$
f'(x)=2x
$$

Finite difference approximation:

$$
f'(x) \approx \frac{f(x+h)-f(x)}{h}
$$

```python
def f(x):
    return x ** 2


def exact_derivative(x):
    return 2 * x


def numerical_derivative(func, x, h=1e-5):
    return (func(x + h) - func(x)) / h

points = np.array([-3, -1, 0, 2, 5], dtype=float)

for point in points:
    print(point, exact_derivative(point), numerical_derivative(f, point))
```

## 2. MSE Derivatives for Simple Linear Regression

$$
\hat{y}_i = wx_i + b
$$

$$
\frac{\partial \mathcal{L}}{\partial w}= -\frac{2}{n}\sum_i x_i(y_i-\hat{y}_i)
$$

$$
\frac{\partial \mathcal{L}}{\partial b}= -\frac{2}{n}\sum_i (y_i-\hat{y}_i)
$$

```python
x_train = np.array([1, 2, 3, 4], dtype=float)
y_train = np.array([3, 5, 7, 9], dtype=float)

w = 0.5
b = 0.0

y_hat = w * x_train + b
errors = y_train - y_hat
mse = np.mean(errors ** 2)

dL_dw = -(2 / len(x_train)) * np.sum(x_train * errors)
dL_db = -(2 / len(x_train)) * np.sum(errors)

mse, dL_dw, dL_db
```

## 3. One Manual Gradient Descent Update

$$
w_{new}=w_{old}-\alpha\frac{\partial \mathcal{L}}{\partial w}
$$

$$
b_{new}=b_{old}-\alpha\frac{\partial \mathcal{L}}{\partial b}
$$

```python
learning_rate = 0.01

w_new = w - learning_rate * dL_dw
b_new = b - learning_rate * dL_db

y_hat_new = w_new * x_train + b_new
mse_new = np.mean((y_train - y_hat_new) ** 2)

w_new, b_new, mse_new
```

## Reflection

Derivatives tell us how loss changes when parameters change. Gradient descent uses this information to update the model.
