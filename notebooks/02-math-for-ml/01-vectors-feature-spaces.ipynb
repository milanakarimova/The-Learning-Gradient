# 01 — Vectors and Feature Spaces Practice

This notebook practices vector representation, dataset matrices, dot products, norms, distances, and feature scaling.

```python
import numpy as np
```

## 1. Feature Map

A real object becomes a vector through a feature map:

$$
\phi : \mathcal{O} \to \mathbb{R}^d
$$

```python
x = np.array([120, 3, 5])

x, x.shape
```

## 2. Dataset Matrix

A dataset with $n$ samples and $d$ features is:

$$
X \in \mathbb{R}^{n \times d}
$$

```python
X = np.array([
    [120, 3, 5],
    [80, 2, 10],
    [200, 5, 2],
    [150, 4, 7],
], dtype=float)

X, X.shape
```

## 3. Dot Product and Linear Model

$$
\hat{y} = w^T x + b
$$

```python
w = np.array([1000, 15000, -3000])
b = 50000

prediction = np.dot(w, x) + b
prediction
```

## 4. Norm and Distance

$$
\|x\|_2 = \sqrt{x_1^2 + x_2^2 + \dots + x_d^2}
$$

$$
d(a,b) = \|a-b\|_2
$$

```python
norm_x = np.linalg.norm(x)
distance = np.linalg.norm(X[0] - X[1])

norm_x, distance
```

## 5. Feature Scaling

$$
z = \frac{x - \mu}{\sigma}
$$

```python
feature_means = X.mean(axis=0)
feature_stds = X.std(axis=0)

X_scaled = (X - feature_means) / feature_stds

X_scaled
```

A vector is not only a mathematical arrow. In Machine Learning, it is a representation of an object.
