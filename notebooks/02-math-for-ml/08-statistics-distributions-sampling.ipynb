# 08 — Statistics, Distributions, and Sampling Practice

This notebook practices descriptive statistics, outlier detection, sampling distributions, bootstrap confidence intervals, covariance, correlation, and distribution shift checks.

```python
import numpy as np
```

## 1. Descriptive Statistics

$$
\bar{x}=\frac{1}{n}\sum_i x_i
$$

$$
s^2=\frac{1}{n-1}\sum_i(x_i-\bar{x})^2
$$

```python
x = np.array([10, 12, 15, 18, 20, 100], dtype=float)

mean = np.mean(x)
median = np.median(x)
variance = np.var(x, ddof=1)
std = np.std(x, ddof=1)

mean, median, variance, std
```

## 2. IQR Outlier Rule

$$
IQR=Q_3-Q_1
$$

Outlier rule:

$$
x<Q_1-1.5IQR \quad \text{or} \quad x>Q_3+1.5IQR
$$

```python
q1 = np.percentile(x, 25)
q3 = np.percentile(x, 75)
iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

outliers = x[(x < lower) | (x > upper)]

q1, q3, iqr, lower, upper, outliers
```

## 3. Sampling Distribution

```python
rng = np.random.default_rng(42)
population = rng.gamma(shape=2.0, scale=2.0, size=10000)

sample_means = []
for _ in range(2000):
    sample = rng.choice(population, size=30, replace=True)
    sample_means.append(np.mean(sample))

sample_means = np.array(sample_means)

np.mean(population), np.mean(sample_means), np.std(sample_means)
```

## 4. Standard Error

$$
SE(\bar{x})=\frac{s}{\sqrt{n}}
$$

```python
sample = rng.choice(population, size=80, replace=False)
standard_error = np.std(sample, ddof=1) / np.sqrt(len(sample))

np.mean(sample), standard_error
```

## 5. Bootstrap Confidence Interval

```python
boot_means = []
for _ in range(5000):
    boot_sample = rng.choice(sample, size=len(sample), replace=True)
    boot_means.append(np.mean(boot_sample))

ci_lower, ci_upper = np.percentile(boot_means, [2.5, 97.5])

ci_lower, ci_upper
```

## 6. Covariance and Correlation

$$
\mathrm{Cov}(X,Y)=\mathbb{E}[(X-\mu_X)(Y-\mu_Y)]
$$

$$
r=\frac{\mathrm{Cov}(X,Y)}{s_Xs_Y}
$$

```python
a = rng.normal(0, 1, size=200)
b = 0.8 * a + rng.normal(0, 0.4, size=200)

cov = np.cov(a, b, ddof=1)[0, 1]
corr = np.corrcoef(a, b)[0, 1]

cov, corr
```

## 7. Train-Test Distribution Shift Check

```python
train = rng.normal(0, 1, size=(500, 3))
test = rng.normal([0.2, 0.0, 1.0], [1.0, 1.0, 1.2], size=(500, 3))

mean_difference = test.mean(axis=0) - train.mean(axis=0)
mean_difference
```
