# 10 — MLE, MAP, and Probabilistic Thinking Practice

This notebook practices Bernoulli MLE, Bernoulli MAP with Beta prior, log-likelihood, Gaussian NLL, and MAP-style regularized objectives.

```python
import numpy as np
```

## 1. Bernoulli Log-Likelihood

$$
\log L(p)=\sum_i x_i\log p+(1-x_i)\log(1-p)
$$

```python
def bernoulli_log_likelihood(p, data, eps=1e-15):
    p = np.clip(p, eps, 1 - eps)
    data = np.array(data)
    return np.sum(data * np.log(p) + (1 - data) * np.log(1 - p))


data = np.array([1, 1, 1, 0, 0])

candidate_ps = np.linspace(0.05, 0.95, 19)
log_likelihoods = np.array([bernoulli_log_likelihood(p, data) for p in candidate_ps])

candidate_ps[np.argmax(log_likelihoods)]
```

## 2. Bernoulli MLE

$$
p_{MLE}=\frac{k}{n}
$$

```python
def bernoulli_mle(data):
    return np.mean(data)

bernoulli_mle(data)
```

## 3. Bernoulli MAP with Beta Prior

If:

$$
p\sim Beta(\alpha,\beta)
$$

then:

$$
p_{MAP}=\frac{\alpha+k-1}{\alpha+\beta+n-2}
$$

```python
def bernoulli_map_beta(data, alpha, beta):
    n = len(data)
    k = np.sum(data)
    return (alpha + k - 1) / (alpha + beta + n - 2)

bernoulli_map_beta(data, alpha=2, beta=5)
```

## 4. Gaussian NLL and MSE

For fixed $\sigma$, Gaussian NLL is proportional to squared error.

```python
def gaussian_nll(y_true, y_pred, sigma=1.0):
    residual = y_true - y_pred
    return np.mean(
        0.5 * np.log(2 * np.pi * sigma ** 2)
        + (residual ** 2) / (2 * sigma ** 2)
    )


def mse(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)


y_true = np.array([3, 5, 7, 9], dtype=float)
y_pred = np.array([2.8, 5.2, 6.9, 9.1], dtype=float)

gaussian_nll(y_true, y_pred), mse(y_true, y_pred)
```

## 5. MAP-Style Regularized Objective

$$
J(w)=\text{data loss}+\lambda\|w\|_2^2
$$

```python
def l2_penalty(w):
    return np.sum(w ** 2)


def map_objective_mse_l2(y_true, y_pred, w, lambda_):
    return mse(y_true, y_pred) + lambda_ * l2_penalty(w)

w = np.array([2.0, -1.0, 0.5])

map_objective_mse_l2(y_true, y_pred, w, lambda_=0.1)
```

## Reflection

Many ML losses are negative log-likelihoods. Regularization can be interpreted as a negative log-prior. This connects optimization with probabilistic modeling.
