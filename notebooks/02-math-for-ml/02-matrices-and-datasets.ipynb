# 02 — Matrices and Datasets Practice

This notebook practices matrix representation of datasets, matrix-vector multiplication, predictions, MSE, augmented design matrices, and centered matrices.

```python
import numpy as np
```

## 1. Feature Matrix and Target Vector

$$
X \in \mathbb{R}^{n \times d}
$$

Rows are samples. Columns are features.

```python
X = np.array([
    [120, 3, 5],
    [80, 2, 10],
    [200, 5, 2],
    [150, 4, 7],
], dtype=float)

y = np.array([180000, 120000, 300000, 240000], dtype=float)

X.shape, y.shape
```

## 2. Batch Linear Prediction

$$
\hat{y} = Xw + b\mathbf{1}
$$

```python
w = np.array([1000, 15000, -3000], dtype=float)
b = 50000

y_pred = X @ w + b

y_pred
```

## 3. MSE

$$
\mathrm{MSE} = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2
$$

```python
errors = y - y_pred
mse = np.mean(errors ** 2)

errors, mse
```

## 4. Augmented Design Matrix

The bias can be absorbed by adding a column of ones.

```python
ones = np.ones((X.shape[0], 1))
X_augmented = np.hstack([ones, X])

X_augmented, X_augmented.shape
```

## 5. Centering and Covariance-like Matrix

If $X$ is centered, then $X^T X$ contains information about how features vary together.

```python
feature_means = X.mean(axis=0)
X_centered = X - feature_means
covariance_like = X_centered.T @ X_centered

feature_means, X_centered, covariance_like
```
