# 09 — Loss Functions and Optimization Practice

This notebook practices regression losses, classification losses, regularization penalties, numerical stability, and gradient descent.

```python
import numpy as np
```

## 1. Regression Losses

$$
\mathrm{MSE}=\frac{1}{n}\sum_i(y_i-\hat{y}_i)^2
$$

$$
\mathrm{MAE}=\frac{1}{n}\sum_i|y_i-\hat{y}_i|
$$

```python
def mse(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)


def mae(y_true, y_pred):
    return np.mean(np.abs(y_true - y_pred))


def huber_loss(y_true, y_pred, delta=1.0):
    residual = y_true - y_pred
    abs_residual = np.abs(residual)
    quadratic = 0.5 * residual ** 2
    linear = delta * (abs_residual - 0.5 * delta)
    return np.mean(np.where(abs_residual <= delta, quadratic, linear))


y_true = np.array([3, 5, 7, 9], dtype=float)
y_pred = np.array([2.8, 5.2, 6.9, 20.0])

mse(y_true, y_pred), mae(y_true, y_pred), huber_loss(y_true, y_pred)
```

## 2. Binary Cross-Entropy

$$
\ell(y,p)=-[y\log(p)+(1-y)\log(1-p)]
$$

```python
def binary_cross_entropy(y_true, p_pred, eps=1e-15):
    p_pred = np.clip(p_pred, eps, 1 - eps)
    loss = -(y_true * np.log(p_pred) + (1 - y_true) * np.log(1 - p_pred))
    return np.mean(loss)


y_binary = np.array([1, 0, 1, 1, 0])
p_binary = np.array([0.9, 0.1, 0.8, 0.4, 0.2])

binary_cross_entropy(y_binary, p_binary)
```

## 3. Softmax Cross-Entropy

```python
def softmax(logits):
    shifted = logits - np.max(logits, axis=1, keepdims=True)
    exp_values = np.exp(shifted)
    return exp_values / np.sum(exp_values, axis=1, keepdims=True)


def softmax_cross_entropy(y_true_indices, logits, eps=1e-15):
    probabilities = softmax(logits)
    n = logits.shape[0]
    correct_probs = probabilities[np.arange(n), y_true_indices]
    correct_probs = np.clip(correct_probs, eps, 1 - eps)
    return np.mean(-np.log(correct_probs))


logits = np.array([[2.0, 0.5, -1.0], [0.2, 1.8, 0.1], [-1.0, 0.5, 2.5]])
y_classes = np.array([0, 1, 2])

softmax(logits), softmax_cross_entropy(y_classes, logits)
```

## 4. Hinge Loss

$$
\ell(y,s)=\max(0,1-ys)
$$

```python
def hinge_loss(y_true, scores):
    margins = y_true * scores
    return np.mean(np.maximum(0, 1 - margins))


y_hinge = np.array([1, -1, 1, -1])
scores = np.array([2.0, -0.3, 0.2, 1.1])

hinge_loss(y_hinge, scores)
```

## 5. Regularization

$$
J(w)=\mathcal{L}(w)+\lambda\Omega(w)
$$

```python
def l1_penalty(w):
    return np.sum(np.abs(w))


def l2_penalty(w):
    return np.sum(w ** 2)


w = np.array([2.0, -1.0, 0.5])

l1_penalty(w), l2_penalty(w)
```

## 6. Gradient Descent on MSE

```python
x_train = np.array([1, 2, 3, 4, 5], dtype=float)
y_train = 2 * x_train + 1

w = 0.0
b = 0.0
learning_rate = 0.01

for step in range(1000):
    y_pred = w * x_train + b
    errors = y_train - y_pred

    dL_dw = -(2 / len(x_train)) * np.sum(x_train * errors)
    dL_db = -(2 / len(x_train)) * np.sum(errors)

    w = w - learning_rate * dL_dw
    b = b - learning_rate * dL_db

w, b
```
