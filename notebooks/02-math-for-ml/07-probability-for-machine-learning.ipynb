This notebook practices expectation, variance, conditional probability, Bayes theorem, likelihood, Gaussian PDF, and Naive Bayes scoring.

```python
import numpy as np
from math import comb, sqrt, pi, exp
```

## 1. Expectation and Variance

$$
\mathbb{E}[X]=\sum_x xP(X=x)
$$

$$
\mathrm{Var}(X)=\mathbb{E}[(X-\mathbb{E}[X])^2]
$$

```python
def expectation(values, probabilities):
    return np.sum(values * probabilities)


def variance(values, probabilities):
    mean = expectation(values, probabilities)
    return np.sum(((values - mean) ** 2) * probabilities)

values = np.array([1, 2, 3, 4, 5, 6], dtype=float)
probabilities = np.ones(6) / 6

expectation(values, probabilities), variance(values, probabilities)
```

## 2. Conditional Probability and Bayes Theorem

$$
P(A|B)=\frac{P(A\cap B)}{P(B)}
$$

$$
P(H|D)=\frac{P(D|H)P(H)}{P(D)}
$$

```python
def conditional_probability(p_a_and_b, p_b):
    return p_a_and_b / p_b


def bayes_theorem(prior, likelihood, evidence):
    return likelihood * prior / evidence

prior = 0.01
likelihood = 0.95
false_positive = 0.05
evidence = likelihood * prior + false_positive * (1 - prior)
posterior = bayes_theorem(prior, likelihood, evidence)

posterior
```

## 3. Bernoulli Likelihood

$$
L(p)=\prod_i p^{x_i}(1-p)^{1-x_i}
$$

$$
\log L(p)=\sum_i x_i\log p + (1-x_i)\log(1-p)
$$

```python
def bernoulli_log_likelihood(p, data):
    data = np.array(data)
    return np.sum(data * np.log(p) + (1 - data) * np.log(1 - p))

rng = np.random.default_rng(42)
data = rng.binomial(1, 0.7, size=20)

candidate_ps = np.linspace(0.1, 0.9, 9)
log_likelihoods = [bernoulli_log_likelihood(p, data) for p in candidate_ps]

candidate_ps, log_likelihoods, candidate_ps[np.argmax(log_likelihoods)]
```

## 4. Gaussian PDF

$$
p(x)=\frac{1}{\sqrt{2\pi\sigma^2}}\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)
$$

```python
def gaussian_pdf(x, mu, sigma):
    coefficient = 1 / (sigma * sqrt(2 * pi))
    exponent = exp(-((x - mu) ** 2) / (2 * sigma ** 2))
    return coefficient * exponent

gaussian_pdf(0, 0, 1)
```

## 5. Tiny Naive Bayes Score

$$
\mathrm{score}(y)=\log P(y)+\sum_j x_j\log P(w_j|y)
$$

```python
def naive_bayes_score(log_prior, word_counts, log_word_probs):
    return log_prior + np.sum(word_counts * log_word_probs)

word_counts = np.array([2, 0, 1, 3])

score_sports = naive_bayes_score(
    np.log(0.6),
    word_counts,
    np.log(np.array([0.30, 0.10, 0.20, 0.40]))
)

score_politics = naive_bayes_score(
    np.log(0.4),
    word_counts,
    np.log(np.array([0.10, 0.40, 0.30, 0.20]))
)

score_sports, score_politics
```
