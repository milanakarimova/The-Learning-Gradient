# 00 — Math Translation Practice

This notebook practices translating basic ML math formulas into Python code.

```python
import numpy as np
```

## 1. Dot Product

$$
w^T x = w_1x_1 + w_2x_2 + \dots + w_dx_d
$$

```python
x = np.array([2, 3, 4])
w = np.array([0.5, 1.0, -0.2])

np.dot(w, x)
```

## 2. Mean, Variance, and Standard Deviation

$$
\bar{x} = \frac{1}{n}\sum_{i=1}^{n}x_i
$$

$$
\mathrm{Var}(x) = \frac{1}{n}\sum_{i=1}^{n}(x_i - \bar{x})^2
$$

```python
values = np.array([10, 20, 30, 40, 50])

np.mean(values), np.var(values), np.std(values)
```

## 3. Standardization

$$
z = \frac{x - \mu}{\sigma}
$$

```python
standardized = (values - np.mean(values)) / np.std(values)
standardized
```

## 4. Mean Squared Error

$$
\mathrm{MSE} = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2
$$

```python
y_true = np.array([3, 5, 2, 7])
y_pred = np.array([2.5, 5.5, 2, 8])

errors = y_true - y_pred
mse = np.mean(errors ** 2)

errors, mse
```
