# 03 — Dot Products and Linear Models Practice

This notebook practices dot products, weighted sums, linear predictions, residuals, and MSE.

```python
import numpy as np
```

## 1. Dot Product

$$
w^T x = w_1x_1 + w_2x_2 + \dots + w_dx_d
$$

```python
x = np.array([120, 3, 5], dtype=float)
w = np.array([1000, 15000, -3000], dtype=float)
b = 50000

manual_prediction = w[0] * x[0] + w[1] * x[1] + w[2] * x[2] + b
dot_prediction = np.dot(w, x) + b

manual_prediction, dot_prediction
```

## 2. Batch Linear Prediction

$$
\hat{y} = Xw + b\mathbf{1}
$$

```python
X = np.array([
    [120, 3, 5],
    [80, 2, 10],
    [200, 5, 2],
    [150, 4, 7],
], dtype=float)

y = np.array([180000, 120000, 300000, 240000], dtype=float)

y_pred = X @ w + b

y_pred
```

## 3. Residuals and MSE

$$
e_i = y_i - \hat{y}_i
$$

$$
\mathrm{MSE} = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2
$$

```python
errors = y - y_pred
mse = np.mean(errors ** 2)

errors, mse
```

## 4. Change One Weight

Changing one weight changes the contribution of that feature.

```python
w_changed = np.array([1200, 15000, -3000], dtype=float)
y_pred_changed = X @ w_changed + b

prediction_difference = y_pred_changed - y_pred
prediction_difference
```

## Reflection

The dot product is a weighted sum, a projection, and a linear score. It is one of the core operations of Machine Learning.
