# 01 — Pandas Basics Practice

This notebook practices basic Pandas operations for Machine Learning: creating a DataFrame, inspecting data, handling missing values, creating features, checking target distribution, and separating X/y.

````python
import pandas as pd
````

## 1. Create a Small Dataset

````python
students = pd.DataFrame({
    "name": ["Aylin", "Rauf", "Leyla", "Murad", "Nigar", "Kamran"],
    "study_hours": [5, 2, 4, None, 6, 1],
    "attendance": [90, 60, 85, 70, 95, 50],
    "passed": [1, 0, 1, 0, 1, 0],
})

students
````

## 2. Inspect the Data

````python
students.head()
````

````python
students.shape, students.columns, students.dtypes
````

````python
students.describe()
````

## 3. Missing Values

````python
students.isna().sum()
````

````python
students["study_hours"] = students["study_hours"].fillna(students["study_hours"].mean())
students
````

## 4. Feature Engineering

````python
students["study_attendance_score"] = students["study_hours"] * students["attendance"]
students
````

## 5. Target Distribution

````python
students["passed"].value_counts()
````

## 6. Separate Features and Target

In Machine Learning, we often use:

```text
X -> features
y -> target
```

````python
X = students[["study_hours", "attendance", "study_attendance_score"]]
y = students["passed"]

X, y
````

## 7. Convert to NumPy

````python
X_np = X.to_numpy()
y_np = y.to_numpy()

X_np, y_np
````
