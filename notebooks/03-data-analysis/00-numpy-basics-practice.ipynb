# 00 — NumPy Basics Practice

This notebook practices the most important NumPy ideas for Machine Learning: arrays, shapes, axis, scaling, vectorized operations, dot products, and MSE.

## 1. Import NumPy

```python
import numpy as np
```

## 2. Create a Feature Matrix

Rows are samples. Columns are features.

```python
X = np.array([
    [50, 2],
    [80, 3],
    [120, 4],
    [150, 5],
], dtype=float)

X
```

```python
X.shape
```

## 3. Feature Means and Standardization

Standardization uses this formula:

$$
z = \frac{x - \mu}{\sigma}
$$

```python
feature_means = np.mean(X, axis=0)
feature_stds = np.std(X, axis=0)

feature_means, feature_stds
```

```python
X_scaled = (X - feature_means) / feature_stds
X_scaled
```

## 4. Linear Model Prediction

A simple linear model can be written as:

$$
\hat{y} = Xw + b
$$

```python
w = np.array([0.3, 10])
b = 5

y_pred = X @ w + b

y_pred
```

## 5. Mean Squared Error

$$
\mathrm{MSE} = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2
$$

```python
y_true = np.array([40, 60, 85, 105])

errors = y_true - y_pred
mse = np.mean(errors ** 2)

errors, mse
```
